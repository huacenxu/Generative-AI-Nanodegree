### Dataset Explanation
I chose the 2023_fashion_trends.csv dataset for this project because:
It contains up-to-date information on 2023 fashion trends, likely more recent than the GPT model's training data.
The rich, text-based format is ideal for OpenAI's language models.
It provides current, industry-relevant insights that the model might not have been originally trained on.
The dataset covers various aspects of fashion trends, including sustainability and color trends.
By using this dataset, I aim to create a chatbot that can provide accurate, timely answers about the latest fashion trends. This demonstrates how custom datasets can enhance AI models with specific, current information, making them more valuable for targeted applications in fashion and trend analysis.

Please note due to the "Insufficient budget available" error with the OpenAI API, I implemented mock data to demonstrate the chatbot's functionality. This allows me to showcase the difference between basic and custom queries, where custom responses utilize the actual fashion trends dataset while basic responses provide generic answers. The mock implementation maintains the core code structure and demonstrates how the system would work with a live API, making it useful for testing and development purposes until API access is restored.

```python
import pandas as pd
import logging
from typing import Dict

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

class MockFashionChatbot:
    def __init__(self):
        self.df = None
        self.trends_data = {}

    def load_data(self, file_path: str):
        self.df = pd.read_csv(file_path)
        self.df['text'] = self.df['Trends'] + " - " + self.df['Source']
        self._process_trends()
        logger.info("Data loaded successfully")

    def _process_trends(self):
        """Process trends data for better contextualization"""
        # Group trends by categories
        for _, row in self.df.iterrows():
            trend = row['Trends'].lower()
            if 'sustainable' in trend:
                self.trends_data.setdefault('sustainable', []).append(row['text'])
            elif '2023' in trend:
                self.trends_data.setdefault('2023', []).append(row['text'])

    def simulate_response(self, question: str, context: str = None) -> str:
        """Basic response without dataset context"""
        basic_responses = {
            "trends_2023": "General 2023 trends include sustainable fashion and bold colors.",
            "sustainable": "General sustainable trends include recycled materials.",
            "default": "Fashion trends vary by season."
        }
        
        if "2023" in question.lower():
            return basic_responses["trends_2023"]
        elif "sustainable" in question.lower():
            return basic_responses["sustainable"]
        return basic_responses["default"]

    def simulate_custom_response(self, question: str, context: list) -> str:
        """Enhanced response using dataset context"""
        if "2023" in question.lower() and self.trends_data.get('2023'):
            trends = ", ".join(self.trends_data['2023'][:3])
            return f"Based on our dataset analysis, specific 2023 trends include: {trends}"
        elif "sustainable" in question.lower() and self.trends_data.get('sustainable'):
            trends = ", ".join(self.trends_data['sustainable'][:3])
            return f"Dataset shows these sustainable trends: {trends}"
        return f"Based on our dataset: {', '.join(context[:3])}"

    def ask_custom(self, question: str) -> str:
        context = self.df['text'].tolist()[:10]
        return self.simulate_custom_response(question, context)

    def ask_basic(self, question: str) -> str:
        return self.simulate_response(question)

def main():
    chatbot = MockFashionChatbot()
    chatbot.load_data("data/2023_fashion_trends.csv")

    questions = [
        "What are the key fashion trends for 2023?",
        "What are some sustainable fashion trends?"
    ]

    print("\nComparing Basic vs Custom Responses:")
    print("-" * 50)
    for q in questions:
        basic = chatbot.ask_basic(q)
        custom = chatbot.ask_custom(q)
        
        print(f"\nQuestion: {q}")
        print(f"Basic: {basic}")
        print(f"Custom (with dataset context): {custom}")
        print("-" * 50)

if __name__ == "__main__":
    main()
```
